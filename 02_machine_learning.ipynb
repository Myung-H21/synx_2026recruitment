## 2.2 机器学习基础概念 —— 综合简答题
### (1) 样本、特征与标签
结合“预测二手房成交价”的例子：
*   **样本**：数据集中的每一条独立记录。比如，某一个具体小区里的一套房子。
*   **特征**：用来描述样本的属性或信息，是模型用来做预测的依据。比如，这套房子的面积、所在楼层、房龄、学区、距离地铁远近等，都可以作为特征。
*   **标签**：样本对应的目标结果（也就是要预测的那个值）。在这个例子中就是这套房子的实际成交价格。

### (2) 训练集、验证集与测试集
*   **各自作用**：
    1.   **训练集**：模型用来“学习”的数据。就像学生平时的课本和习题，模型通过它来寻找规律（训练参数）。
    2.   **验证集**：用来“调参”和“选模型”的数据。就像学生的模拟考，用来检验哪个模型当前表现最好，或者调整模型的超参数（比如调整正则化强度），防止模型在训练集上死记硬背。
    3.   **测试集**：用来“最终考核”的数据。就像真正的高考，模型在训练和调参阶段绝对不能碰它TT，用来真实评估模型在未知数据上的表现（泛化能力）。
*   **为什么不能根据测试集表现反复调整模型？**
  这就好比学生提前偷看了高考卷子并把答案背了下来。如果你根据测试集的表现去修改模型，测试集的信息就会泄露给模型，导致最终评估出来的分数虚高（过拟合测试集），无法真实反映模型在现实真实世界中的表现。

### (3) 监督学习与无监督学习
*   **监督学习**：数据既有特征，又有明确的“标签（答案）”。模型的任务是学习从特征到标签的映射关系。
  >实际任务举例：垃圾邮件识别（给模型看大量标注了“是垃圾邮件”或“不是垃圾邮件”的邮件，让它学会自己分类）
*   **无监督学习**：数据只有特征，没有标签（没有标准答案）。模型的任务是发现数据内部的结构、规律或分组。
  >实际任务举例：电商用户群体划分（给模型看用户的购买记录、浏览行为，模型自动把消费习惯相似的人聚成一类

### (4) 回归、分类与聚类
1.  **根据房屋属性预测房价** —— **回归**
   理由：预测的目标（房价）是连续型数值。
2.  **判断一封邮件是不是垃圾邮件** —— **分类**
   理由：预测的目标是离散的类别（是/否，属于二分类问题）。
3.  **根据用户行为自动把用户分成若干群体** —— **聚类**
   理由：事先并没有定义好有哪些群体标签，属于无监督学习，让算法根据相似度自动抱团。

### (5) 数据预处理
列举三种常见的预处理操作及解决的问题：
1.  **缺失值处理（填充或删除）**：
    *  解决什么问题：原始数据中某些特征有空缺，大多数算法无法直接处理含空值的数据，会导致报错或预测偏差。
2.  **特征缩放（归一化/标准化）**：
    *   解决什么问题：不同特征的量纲差异巨大（比如年龄是0-100，而收入是0-100000）。这会导致距离计算（如KNN）或梯度下降算法失效，模型训练缓慢或偏向数值大的特征。缩放后能让不同特征在同等尺度下发挥作用。
3.  **类别特征编码（如独热编码 One-Hot）**：
    *   解决什么问题：机器学习模型底层是数学计算，无法直接识别“红色”、“蓝色”这样的文本字符。编码可以将其转化为数字（如0和1），让模型能够处理。


### (6) 过拟合与泛化
*   **什么是过拟合**：模型在训练集上表现得极其完美，甚至把数据里的“噪音（随机误差）”都当成了规律记了下来，但一到新数据（测试集）上就一塌糊涂。也就是“死记硬背”了，缺乏举一反三的能力
*   **什么是模型的泛化能力**：模型在“没见过的”新数据上表现良好的能力。这才是模型真正有用的体现
*   **训练集99%，测试集60%，可能的原因**：
    1.  **严重过拟合**：模型太复杂（参数太多），把训练集背下来了
    2.  **数据泄露**：不小心把测试集的数据混进训练集里了
    3.  **数据分布不一致**：训练集和测试集的规律完全不同（比如用2020年的房价数据训练，去预测2024年的房价）
    4.  **测试集样本太少或存在极端情况**，导致评估结果偶然性太大

### (7) 模型评估指标
*   **指标含义**：
    *   **Accuracy（准确率）**：所有预测对的样本数 / 总样本数（整体猜对的概率）
    *   **Precision（精确率）**：预测为正样本中，真正是正样本的比例（宁缺毋滥，怕误报）
    *   **Recall（召回率）**：真正是正样本中，被成功预测出来的比例（宁可错杀一千，不可放过一个，怕漏报）
    *   **F1-score**：Precision和Recall的调和平均数。当两者发生冲突时，用来综合评估
    *   **MSE（均方误差）**：预测值与真实值差值的平方的平均数。常用于回归问题，对误差大的预测惩罚极重
    *   **MAE（平均绝对误差）**：预测值与真实值差值绝对值的平均数。常用于回归问题，更能反映预测误差的真实大小
*   **场景分析：“严重疾病筛查”，漏掉一个病人的代价极大。**
    *   更关注 **Recall（召回率）**
    *   原因：漏诊（本来是病人，模型预测没病，这叫假阴性）会危及生命；而误诊（本来没病，预测有病，叫假阳性）仅仅是让人多做几项检查。所以我们要尽可能把所有真正的病人都找出来，Recall越高，漏诊越少

## 2.3 线性模型、损失函数、梯度下降与 Softmax —— 理论综合题

### （1）线性回归

给定一个具有 \(d\) 个特征的样本：
$$
x = [x_1, x_2, \dots, x_d]^T
$$

线性回归模型定义为：
$$
\hat{y} = w^T x + b
$$

**回答：**
1.  **\(w\) 与 \(b\) 分别代表什么？**
    *   **\(w\)（权重向量）**：代表各个特征对预测结果的影响大小和方向。\(w = [w_1, w_2, \dots, w_d]^T\)，其中 \(w_i\) 表示第 \(i\) 个特征 \(x_i\) 的贡献
    *   **\(b\)（偏置项/截距）**：表示当所有输入特征均为 0 时模型的基准预测值。引入 \(b\) 让模型不强制穿过原点，增加了拟合的灵活性
2.  **为什么称之为“线性”回归？**
    *   从模型输出看：预测值 \(\hat{y}\) 是输入特征 \(x\) 的**线性组合**。在二维空间是一条直线，三维空间是一个平面，高维空间是超平面
    *   从参数看：模型关于未知参数 \(w\) 和 \(b\) 是线性的（没有相互乘积或非线性变换）
3.  **如果要预测房价，请简述该模型可能用到哪些特征？**
    *   房屋面积、居室数量、房龄、距离市中心的距离、所在楼层、朝向、周边配套（如是否有地铁、学区）等


### （2）均方误差 MSE

对于 \(n\) 个样本，定义：
$$
L = \frac{1}{n} \sum_{i=1}^n (\hat{y}_i - y_i)^2
$$

**回答：**
1.  **为什么回归任务中经常使用 MSE？**
    *   数学性质好：MSE 是连续可导的凸函数，优化时容易求梯度，且能保证全局最优解（在线性回归下）
    *   物理意义明确：对应欧氏距离的平方，衡量的是预测值与真实值的平均偏离程度
2.  **当预测值与真实值误差大时，平方项会带来什么影响？**
    *   **放大惩罚**：平方操作会让较大的误差产生极大的损失值（如误差为 10，损失为 100）。这会迫使模型优先去拟合那些误差极大的样本
3.  **MSE 是否容易受到离群点影响？为什么？**
    *   **极易受到影响**。因为离群点（异常值）的误差通常很大，平方后损失会呈指数级放大，模型为了降低整体 Loss，可能会过度向离群点偏移，导致整体拟合效果变差

### （3）线性回归的梯度推导

设：
$$
\hat{y}_i = w^T x_i + b
$$
损失函数：
$$
L = \frac{1}{n} \sum_{i=1}^n (\hat{y}_i - y_i)^2
$$

**回答：**
根据链式法则：
$$
\frac{\partial L}{\partial \hat{y}_i} = \frac{1}{n} \cdot 2(\hat{y}_i - y_i) = \frac{2}{n}(\hat{y}_i - y_i)
$$

对 \(w\) 求导（\(x_i\) 为列向量，结果也为列向量）：
$$
\frac{\partial \hat{y}_i}{\partial w} = x_i
$$
$$
\frac{\partial L}{\partial w} = \sum_{i=1}^n \frac{\partial L}{\partial \hat{y}_i} \frac{\partial \hat{y}_i}{\partial w} = \frac{2}{n} \sum_{i=1}^n (\hat{y}_i - y_i) x_i
$$

对 \(b\) 求导：
$$
\frac{\partial \hat{y}_i}{\partial b} = 1
$$
$$
\frac{\partial L}{\partial b} = \sum_{i=1}^n \frac{\partial L}{\partial \hat{y}_i} \frac{\partial \hat{y}_i}{\partial b} = \frac{2}{n} \sum_{i=1}^n (\hat{y}_i - y_i)
$$


### （4）完成一次梯度下降更新

给定样本：\(x = [1, 2]^T, y = 5\)
初始参数：\(w = [0, 1]^T, b = 0\)
优化目标函数：\(L = (\hat{y} - y)^2\)
学习率：\(\eta = 0.1\)

**回答：**
1.  **当前预测值 \(\hat{y}\)**：
$$
\hat{y} = w^T x + b = 0 \cdot 1 + 1 \cdot 2 + 0 = 2
$$
2.  **当前 Loss**：
$$
L = (2 - 5)^2 = (-3)^2 = 9
$$
3.  **\(\frac{\partial L}{\partial w}, \frac{\partial L}{\partial b}\) 的值**：
    先对预测值求导：
$$
\frac{\partial L}{\partial \hat{y}} = 2(\hat{y} - y) = 2(2 - 5) = -6
$$
    然后求 \(w\) 的梯度：
$$
\frac{\partial L}{\partial w} = \frac{\partial L}{\partial \hat{y}} \cdot x = -6 \cdot \begin{bmatrix} 1 \\ 2 \end{bmatrix} = \begin{bmatrix} -6 \\ -12 \end{bmatrix}
$$
    求 \(b\) 的梯度：
$$
\frac{\partial L}{\partial b} = \frac{\partial L}{\partial \hat{y}} \cdot 1 = -6
$$
4.  **完成一次梯度下降后的新参数**：
$$
w_{new} = w - \eta \frac{\partial L}{\partial w} = \begin{bmatrix} 0 \\ 1 \end{bmatrix} - 0.1 \begin{bmatrix} -6 \\ -12 \end{bmatrix} = \begin{bmatrix} 0.6 \\ 2.2 \end{bmatrix}
$$
$$
b_{new} = b - \eta \frac{\partial L}{\partial b} = 0 - 0.1 \cdot (-6) = 0.6
$$



### （5）Batch、SGD 与 Mini-Batch

**回答：**
依次对比三个梯度下降变体：
1.  **每次更新使用多少数据**：
    *   **Batch GD（全量）**：使用全部 \(n\) 个样本
    *   **SGD（随机）**：每次仅使用 1 个样本
    *   **Mini-Batch**：每次使用一小批数据（如 32、64、128 等）
2.  **单次计算开销**：
    *   Batch 开销最大，SGD 开销极小，Mini-Batch 开销适中
3.  **梯度稳定性**：
    *   Batch 梯度最稳定（真实梯度方向）；SGD 梯度极度不稳定（方差大，震荡严重）；Mini-Batch 介于两者之间，较为稳定
4.  **收敛速度**：
    *   单次迭代 Batch 最慢；SGD 最快（但震荡大可能不收敛）；Mini-Batch 收敛速度较快且稳定
5.  **是否适合 GPU 并行**：
    *   Batch 和 Mini-Batch 适合（利用矩阵运算高度并行）；SGD 不适合（单样本无法发挥 GPU 优势）
6.  **实际深度学习中为什么常用 Mini-Batch**：
    *   兼顾了**计算效率**（利用 GPU 矩阵运算并行）和**收敛稳定性**（相比 SGD 震荡小），同时内存消耗可控（Batch 太大容易导致 OOM）

### （6）Softmax 回归
Softmax 定义：
$$
\hat{y}_i = \frac{e^{z_i}}{\sum_{j=1}^3 e^{z_j}}, \quad i = 1, 2, 3
$$

**回答：**
1.  **Softmax 输出为什么可以看作概率分布？**
    *   因为对于任意 \(i\)，\(e^{z_i} > 0\)，所以 \(0 < \hat{y}_i < 1\)。
    *   且所有类别的输出值之和为：
$$
\sum_{i=1}^3 \hat{y}_i = \frac{\sum_{i=1}^3 e^{z_i}}{\sum_{j=1}^3 e^{z_j}} = 1
$$

2.  **为什么要用指数函数 \(e^{z_i}\)？**
    *   保证所有输出值为**非负**（概率不能为负）
    *   指数函数具有**放大差异**的作用，较大的 logit 
    *   求导性质极其优良（导数包含自身）
3.  **当某个 logit 很大时，Softmax 输出会发生什么变化？**
    *   发生“**赢者通吃（Winner-takes-all）**”现象。如果 \(z_k\) 远大于其他值，则 \(e^{z_k}\) 在分母中占主导，导致 \(\hat{y}_k \approx 1\)，其他类别概率趋近于 0
4.  **为什么实际实现 Softmax 时要处理数值稳定性（如 max 技巧）？**
    *   如果 \(z_i\) 过大（如 1000），\(e^{1000}\) 会导致计算机浮点数溢出，变成 `inf`
    *   **Max 技巧**：在分子分母同乘 \(e^{-\max(z)}\)，即计算：
$$
\hat{y}_i = \frac{e^{z_i - \max(z)}}{\sum_j e^{z_j - \max(z)}}
$$
    这样最大的指数项变为 \(e^0=1\)，避免了溢出


### （7）Softmax + Cross Entropy 推导

**回答：**
1.  **什么是交叉熵？**
    *   交叉熵衡量两个概率分布（真实分布 \(y\) 和预测分布 \(\hat{y}\)）之间的差异。公式为：
$$
L = -\sum_{i=1}^C y_i \log(\hat{y}_i)
$$
>其中 \(y\) 为 one-hot 标签，\(C\) 为类别数
2.  **推导交叉熵关于 logits \(z\) 的梯度，并证明**
$$
\frac{\partial L}{\partial z_i} = \hat{y}_i - y_i
$$
*  **推导过程**：
    对于某个 logit \(z_k\)，求偏导 \(\frac{\partial L}{\partial z_k}\)
    根据链式法则：
$$
\frac{\partial L}{\partial z_k} = -\sum_{i=1}^C y_i \frac{\partial \log(\hat{y}_i)}{\partial z_k} = -\sum_{i=1}^C \frac{y_i}{\hat{y}_i} \frac{\partial \hat{y}_i}{\partial z_k}
$$

我们需要计算 \(\frac{\partial \hat{y}_i}{\partial z_k}\)。由于 \(\hat{y}_i = \frac{e^{z_i}}{\sum_j e^{z_j}}\)，分两种情况讨论：
*   **当 \(k = i\) 时**：
$$
\frac{\partial \hat{y}_i}{\partial z_i} = \frac{e^{z_i}\sum_j e^{z_j} - e^{z_i}e^{z_i}}{(\sum_j e^{z_j})^2} = \hat{y}_i - \hat{y}_i^2 = \hat{y}_i(1 - \hat{y}_i)
$$
 *   **当 \(k \neq i\) 时**：
$$
\frac{\partial \hat{y}_i}{\partial z_k} = \frac{0 - e^{z_i}e^{z_k}}{(\sum_j e^{z_j})^2} = -\hat{y}_i \hat{y}_k
$$

        将上述结果代回梯度公式：
$$
\frac{\partial L}{\partial z_k} = -\frac{y_k}{\hat{y}_k} \hat{y}_k(1 - \hat{y}_k) - \sum_{i \neq k} \frac{y_i}{\hat{y}_i} (-\hat{y}_i \hat{y}_k)
$$
$$
= -y_k(1 - \hat{y}_k) + \sum_{i \neq k} y_i \hat{y}_k
$$
$$
= -y_k + y_k \hat{y}_k + \hat{y}_k \sum_{i \neq k} y_i
$$

由于 \(y\) 是 one-hot 向量，总和 \(\sum_{i=1}^C y_i = 1\)，因此 \(\sum_{i \neq k} y_i = 1 - y_k\)。代入得：
$$
\frac{\partial L}{\partial z_k} = -y_k + y_k \hat{y}_k + \hat{y}_k(1 - y_k) = -y_k + y_k \hat{y}_k + \hat{y}_k - \hat{y}_k y_k = \hat{y}_k - y_k
$$


>化简后，梯度向量形式为 \(\nabla_z L = \hat{y} - y\)。这一结果极其简洁，也是深度学习框架中 Softmax 结合交叉熵作为损失函数时，反向传播计算如此高效的原因

# 2.4 PyTorch MLP 回归实验 —— 从 Tensor 手搓到 nn.Module
## (1) MLP 基础理解
MLP（多层感知机）：由输入层、若干隐藏层和输出层组成的前馈神经网络，层间全连接，隐藏层通过非线性激活函数增强表达能力。
1.  **线性模型存在的问题与 MLP 的引入**：
线性模型只能拟合线性关系（直线或平面）。现实中的数据往往存在复杂的非线性关系。引入 MLP（多层感知机）后，通过**隐藏层**和**非线性激活函数**，理论上可以逼近任意复杂的连续函数
2.  **各层的作用**：
    *   **输入层**：接收原始特征数据。
    *   **隐藏层**：进行特征提取、非线性映射和数据降维/升维，学习数据内部的高级表示。
    *   **输出层**：将学到的特征映射到最终的预测结果（回归任务输出 1 个连续值）。
3.  **多层神经网络不使用激活函数**：无论多少层，本质上仍然是一个线性变换。因为 \(W_2(W_1 x) = (W_2 W_1)x\)，多个线性层叠加等价于一层线性层，模型将失去学习非线性特征的能力。
4.  **ReLU 的定义及对比 Sigmoid**：
    *   定义：\(f(x) = \max(0, x)\)。
    *   优点：计算极快；在正区间解决了**梯度消失**问题；具有稀疏性（负半轴输出为0）
   >Sigmoid 缺点：容易导致梯度消失（两端饱和）；输出不是零均值（导致收敛慢）；包含指数运算，计算慢。
5.  **回归任务输出层不需要 ReLU**：回归任务的标签 \(y\) 范围可能是正数、负数或零。ReLU 会将输出截断为非负数（\(\ge 0\)），这将严重限制模型的表达能力。输出层通常保持线性（不加激活函数）。
6.  **训练过程描述**：
    *   **Forward（前向传播）**：数据从输入层经过各个隐藏层计算，最终得到预测值。
    *   **Loss（计算损失）**：比较预测值和真实值，计算它们之间的误差（如 MSE）。
    *   **Backward（反向传播）**：根据微积分的链式法则，从输出层往回推导，计算每个参数对误差的“贡献（梯度）”。
    *   **参数更新**：根据梯度方向，稍微调整参数（\(w = w - lr \times grad\)），使下一次预测的误差更小。


```python
# A
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# 0. 数据准备
# 读取 CSV 文件 
# 假设文件名为 mlp_regression_dataset.csv
df = pd.read_csv('mlp_regression_dataset.csv')

# 提取特征 x1~x10 和标签 y，忽略 id
X = df[[f'x{i}' for i in range(1, 11)]].values
y = df['y'].values.reshape(-1, 1) # 形状调整为 (200, 1)

# 特征标准化 (非常重要，有助于10维特征的模型收敛)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 转换为 Tensor
X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

# 1. 初始化参数
torch.manual_seed(42)
# 输入维度 10, 隐藏层维度 32 (可以自己调整), 输出维度 1
D_in, H, D_out = 10, 32, 1

# 使用 Kaiming 初始化 (适合 ReLU)
w1 = torch.randn(D_in, H, dtype=torch.float32) * np.sqrt(2.0 / D_in)
b1 = torch.zeros(H, dtype=torch.float32)
w2 = torch.randn(H, D_out, dtype=torch.float32) * np.sqrt(2.0 / H)
b2 = torch.zeros(D_out, dtype=torch.float32)

learning_rate = 0.05
epochs = 500 # 回归问题通常需要多跑几个 epoch

# 2. 定义网络结构 (前向传播函数)
def forward(x, w1, b1, w2, b2):
    z1 = torch.matmul(x, w1) + b1
    h = torch.relu(z1)
    y_pred = torch.matmul(h, w2) + b2
    return y_pred, z1, h

# 3. 定义损失函数 (MSE)
def mse_loss(y_pred, y_true):
    n = y_true.shape[0]
    # 均方误差：预测值与真实值差的平方和，除以样本数
    loss = ((y_pred - y_true) ** 2).sum() / (2 * n)
    return loss

# 4. 训练循环
losses = []

for epoch in range(epochs):
    # 1. 前向传播
    y_pred, z1, h = forward(X_tensor, w1, b1, w2, b2)
    
    # 2. 计算损失
    loss = mse_loss(y_pred, y_tensor)
    losses.append(loss.item())
    
    # 3. 手动反向传播
    N = X_tensor.shape[0]
    
    # dL/dy_pred
    dy_pred = (y_pred - y_tensor) / N
    
    # dL/dw2 = h^T @ dy_pred
    dw2 = torch.matmul(h.T, dy_pred)
    db2 = dy_pred.sum(dim=0)
    
    # dL/dh = dy_pred @ w2^T
    dh = torch.matmul(dy_pred, w2.T)
    
    # ReLU 导数
    dz1 = dh * (z1 > 0).float()
    
    # dL/dw1 = x^T @ dz1
    dw1 = torch.matmul(X_tensor.T, dz1)
    db1 = dz1.sum(dim=0)
    
    # 4. 手动更新参数
    with torch.no_grad():
        w1 -= learning_rate * dw1
        b1 -= learning_rate * db1
        w2 -= learning_rate * dw2
        b2 -= learning_rate * db2
        
    if (epoch + 1) % 200 == 0:
        print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.6f}")

# 5. 可视化结果 (回归问题不能画决策边界，画真实值vs预测值)
plt.figure(figsize=(12, 5))

# 左图：Loss 变化
plt.subplot(1, 2, 1)
plt.plot(losses)
plt.title("Training Loss (Version A)")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

# 右图：真实值 vs 预测值散点图
plt.subplot(1, 2, 2)
with torch.no_grad():
    y_final_pred, _, _ = forward(X_tensor, w1, b1, w2, b2)
    y_final_pred = y_final_pred.numpy()
    
plt.scatter(y, y_final_pred, alpha=0.6, color='blue')
# 画一条 y=x 的参考线，点越贴近这条线，预测越准确
min_val, max_val = min(y.min(), y_final_pred.min()), max(y.max(), y_final_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--')
plt.title("True vs Predicted (Version A)")
plt.xlabel("True y")
plt.ylabel("Predicted y")

plt.tight_layout()
plt.show()
```

```python
# B
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# 0. 数据准备 (与版本A完全一致)
df = pd.read_csv('mlp_regression_dataset.csv')
X = df[[f'x{i}' for i in range(1, 11)]].values
y = df['y'].values.reshape(-1, 1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

# 1. 定义模型
class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        
    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        return out

model = MLP(input_dim=10, hidden_dim=32, output_dim=1)

# 2. 定义损失函数和优化器
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)

# 3. 训练循环
epochs = 500
losses_b = []

for epoch in range(epochs):
    optimizer.zero_grad()
    y_pred = model(X_tensor)
    loss = criterion(y_pred, y_tensor)
    losses_b.append(loss.item())
    loss.backward()
    optimizer.step()
    
    if (epoch + 1) % 200 == 0:
        print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.6f}")

# 4. 可视化结果
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(losses_b)
plt.title("Training Loss (Version B)")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.subplot(1, 2, 2)
with torch.no_grad():
    y_final_pred = model(X_tensor).numpy()
    
plt.scatter(y, y_final_pred, alpha=0.6, color='green')
min_val, max_val = min(y.min(), y_final_pred.min()), max(y.max(), y_final_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--')
plt.title("True vs Predicted (Version B)")
plt.xlabel("True y")
plt.ylabel("Predicted y")

plt.tight_layout()
plt.show()
```

### 2. Tensor 手搓版本中，最容易写错的地方是什么？
1.  **矩阵转置（`.T`）漏写或写错位置**：
    *   在推导 `dw1 = X^T @ dz1` 和 `dw2 = h^T @ dy_pred` 时，如果忘记了转置，不仅维度会报错，逻辑也是错的。手推链式法则时，矩阵的左右顺序和转置极易搞混
2.  **ReLU 的导数写错**：
    *   ReLU 在 \(x > 0\) 时导数为 1，\(x \le 0\) 时导数为 0。手写代码 `dz1 = dh * (z1 > 0).float()` 时，很容易忘记加 `.float()` 导致布尔值无法参与乘法，或者弄错判断条件
3.  **忘记加 `with torch.no_grad():`**：
    *   在最后更新参数 `w1 -= lr * dw1` 时，如果不加 `no_grad` 上下文管理器，PyTorch 会把更新操作也记录到计算图中，导致内存爆炸或报错
4.  **MSE 求导时的系数（1/2 或 1/N）**：
    *   Loss 公式是 \(MSE = \frac{1}{N}\sum(y_{pred} - y)^2\)。求导后前面会多出一个 \(2\)。为了消掉这个 \(2\)，通常把 Loss 定义为 \(\frac{1}{2N}\sum(...)^2\)。如果在手写 `dy_pred` 时系数写错（比如多乘了2或少除了N），会导致梯度爆炸或收敛极慢

### 3. `nn.Module` 和 `torch.optim` 帮你封装掉了哪些工作？
**`nn.Module` 的封装：**
*   **参数管理**：它像一个大管家，自动把层里的 `w` 和 `b` 收集起来。你只需调用 `model.parameters()`，就能一键获取所有需要更新的参数，不用像版本A那样手动去传 `w1, b1, w2, b2`。
*   **运算逻辑**：`nn.Linear` 自动完成了 `x @ W.T + b` 的矩阵运算，并且帮你处理好了维度的映射，不用你再手写 `torch.matmul`。
*   **默认初始化**：帮你自动做了 Kaiming 或 Xavier 等科学的权重初始化。
**`torch.optim` 的封装：**
*   **自动更新参数**：`optimizer.step()` 自动执行了 `w = w - lr * w.grad`。你完全不需要手动写更新逻辑。
*   **梯度清零**：`optimizer.zero_grad()` 自动遍历所有参数，把上一轮的梯度清零，防止梯度累加。
*   **进阶优化算法（隐藏福利）**：如果你想用 Adam 优化器，只需要改一行代码 `torch.optim.Adam`，它底层包含了复杂的自适应学习率和动量计算，手写几乎不可能。

**（补充：`loss.backward()` 的封装）**
`loss.backward()` **自动完成了最复杂的链式法则求导**。手写的几十行求导公式（包括矩阵转置、ReLU求导），PyTorch 的 Autograd 引擎在底层瞬间就算完了，这也是为什么版本B不容易出错的核心原因。

# （7）过拟合和欠拟合思考
## 1. 这通常说明发生了什么？
**发生了过拟合（Overfitting）。** 模型过度学习了训练集中的细节甚至噪声，导致在训练集上表现完美，但丧失了泛化能力，在未见过的测试集上表现变差

## 2. 为什么训练集上越来越好，不代表模型真的越来越好？
训练 Loss 下降可能只是模型在“死记硬背”训练数据（包括噪声）。我们训练模型的最终目的是**泛化**（解决未知的新问题），而不是在已知的题库里拿满分

## 3. 如果让你改进，你会尝试哪些办法？
*   **早停法**：在测试集 Loss 开始上升时停止训练
*   **降低模型复杂度**：减少隐藏层数或神经元个数
*   **加入正则化**：使用 L2 正则化）或 Dropout 层
*   **增加数据**：收集更多数据或进行数据增强

## 2.5 正则化与 Dropout
### (1) L1 与 L2 正则化

*   **什么是正则化？**
    正则化是一种防止模型“死记硬背”（过拟合）的技术。通俗来说，就是在训练模型时，除了要求模型把训练数据预测准之外，还额外施加一种“约束”或“惩罚”，限制模型参数（权重）的大小，让模型变得更简单、更平滑，从而提高它在未见过的数据（测试集）上的表现

*   **L1 Regularization 与 L2 Regularization 分别在 Loss 中加入了什么？**
    *   **L1 正则化**：在原来的损失函数（Loss）基础上，加入了**所有权重绝对值之和**（即 L1 范数）
    *   **L2 正则化**：在原来的损失函数基础上，加入了**所有权重平方和**（即 L2 范数）

*   **它们为什么可能缓解过拟合？**
    过拟合往往是因为模型为了迎合包含噪声的训练数据，导致某些特征的权重变得非常大。加入正则化项后，模型为了最小化总损失，必须限制权重变大。**权重变小意味着模型的输出函数变得更平滑，对输入数据的微小变化不那么敏感**，从而放弃去拟合那些局部的噪声，转而学习数据中更普遍、更本质的规律，这就缓解了过拟合

*   **L1 与 L2 对模型参数可能产生什么不同影响？**
    *   **L1（Lasso）**：倾向于让很多不重要的特征的权重**直接变成 0**。这会产生**稀疏矩阵**，相当于自动做了一次特征选择（只保留少数重要的特征）
    *   **L2（Ridge）**：倾向于让所有的权重都**变小、趋于均匀**，但一般不会直接变成 0。它能让模型更稳定，尤其是在特征之间存在相关性时，L2 会让权重分配更平滑


### (2) Dropout

*   **1. Dropout 的基本思想是什么？**
    在神经网络的**训练阶段**，每次前向传播时，按照一定的概率（比如 50%）**随机“丢弃”（即暂时将其输出置为 0）**一部分神经元。这就好像每次训练时都在一个残缺的网络上进行

*   **2. 为什么随机“丢弃”一部分神经元反而可能提升泛化能力？**
    主要有两个原因：
    *   **防止神经元共适应**：如果不加 Dropout，某些神经元可能会过度依赖其他特定神经元的存在。随机丢弃后，神经元不能指望特定的“队友”，必须自己学会提取有用的特征，这增强了单个神经元的鲁棒性
    *   **集成学习效应**：每次 Dropout 相当于生成了一个不同的子网络（相当于训练了成百上千个不同的模型）。在测试时，相当于把所有这些子网络的预测结果综合起来（类似集成学习），从而有效降低方差，提升泛化能力

*   **3. Dropout 在训练阶段和测试阶段的行为是否相同？**
    **不相同**
    *   **训练阶段**：随机激活/丢弃神经元，通常还会对保留的神经元输出进行缩放（比如除以保留概率 \(1-p\)），以保持输出的期望值不变。
    *   **测试阶段**：**不使用** Dropout，所有神经元都参与工作（权重全部激活），以利用模型完整的计算能力。

*   **4. 如果忘记在测试时切换到正确模式，可能产生什么问题？**
    如果在测试（推理）时依然开启了 Dropout，那么每次输入相同的数据，模型都会随机丢弃不同的神经元，导致**每次预测的结果都不一样**（输出具有随机性）。同时，因为测试时少了一部分神经元的参与，导致整体输出的数值期望会变小（产生偏差），这会**极大地降低测试的准确率和稳定性**
