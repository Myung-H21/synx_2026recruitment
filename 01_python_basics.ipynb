```python
# 1.1 Python 基础语法——活动兴趣分级器
# Step 1: 定义获取兴趣等级的函数
def get_interest_level(score):
    """
    根据输入的评分（0-100），返回对应的等级和含义
    """
    if 90 <= score <= 100:
        return "A (强烈推荐)"
    elif 80 <= score < 90:
        return "B (推荐)"
    elif 70 <= score < 80:
        return "C (可以考虑)"
    elif 60 <= score < 70:
        return "D (兴趣一般)"
    else:
        return "E (暂不推荐)"

# Step 2 & Step 3: 循环录入 + 异常处理
print("--- 欢迎使用活动兴趣分级器 ---")

# 获取活动数量 (n)，并进行验证 (Step 3)
while True:
    try:
        n_input = input("请输入活动数量：").strip()
        n = int(n_input)
        if n > 0:
            break
        else:
            print(" 错误：活动数量必须是大于0的正整数，请重新输入。\n")
    except ValueError:
        print(" 错误：输入无效，请输入一个正整数（如 3）。\n")

# 初始化用于统计的变量
total_score = 0
max_score = -1
max_activity_name = ""
# 用于保存所有活动信息的列表（可选，如果后续需要）
activities_data = []

# 循环 n 次获取活动信息
for i in range(1, n + 1):
    print(f"\n--- 录入第 {i} 个活动 ---")
    
    # 1. 活动名称验证
    while True:
        activity_name = input("请输入活动名称：").strip()
        if activity_name:  # 非空校验
            break
        else:
            print(" 错误：活动名称不能为空，请重新输入。")

    # 2. 评分验证
    while True:
        try:
            score_input = input("请输入兴趣评分：").strip()
            score = float(score_input)
            if 0 <= score <= 100:
                break
            else:
                print("错误：评分必须在 0 到 100 之间，请重新输入。")
        except ValueError:
            print(" 错误：输入无效，请输入数字（整数或小数均可）。")

    # 3. 调用函数获取等级并输出
    level = get_interest_level(score)
    print(f"兴趣等级：{level}")
    
    # 4. 更新统计数据
    activities_data.append({"name": activity_name, "score": score, "level": level})
    total_score += score
    
    # 判断最高分（如果有同分，保留先录入的）
    if score > max_score:
        max_score = score
        max_activity_name = activity_name

# 统计与最终输出
print("\n" + "="*30)
average_score = total_score / n
print(f"平均兴趣分：{average_score:.2f}")
print(f"最高分：{max_score:.2f}")
print(f"最高分活动：{max_activity_name}")
print("="*30)
```

# “最后想一想” —— 为什么要定义函数？
1. 代码复用： 将“判断等级”的逻辑封装到 get_interest_level() 中。后续如果有其他活动、或者程序的其他部分也需要判断等级，直接调用这个函数即可，不需要把那一大串 if-elif-else 复制粘贴好几遍。

2. 便于维护： 如果以后学校修改了评级标准（比如把 60 分及格改成 70 分及格），只需要在函数里面改一次，整个程序都会生效，极大地减少了出错的风险。

3. 结构清晰： 主程序只需要关心“获取数据 -> 调用函数 -> 统计结果”，而不需要被复杂的判断逻辑弄乱。


```python
# 1.2 Python 数据结构——活动评论关键词探测器
def process_text():
    #  1. 输入与校验部分 
    try:
        n = int(input("请输入文本行数 n: "))
        if n < 0:
            print("行数不能为负数，请重新运行。")
            return
    except ValueError:
        print("输入格式错误，请输入整数。")
        return

    text_lines = []
    for i in range(n):
        text_lines.append(input(f"请输入第 {i+1} 行文本: "))

    try:
        k = int(input("请输入需要输出的 Top K 词数: "))
        if k < 0:
            print("Top K 不能为负数，请重新运行。")
            return
    except ValueError:
        print("输入格式错误，请输入整数。")
        return

    stop_words = set()
    if k > 0:
        while True:
            stop_words_input = input("请输入停用词（以空格分隔）: ")
            stop_words_list = stop_words_input.split()
            
            # 校验输入的停用词数量是否等于 k
            if len(stop_words_list) != k:
                print("输入有误，停用词数量与 K 不匹配，请重新输入这一行。")
                continue
            else:
                # 停用词统一转换为小写，使用 set 存储提高查找效率
                stop_words = set([word.lower() for word in stop_words_list])
                break
    else:
        # k = 0 时，不输入停用词
        pass

    # 2. 文本预处理与分词 
    # 将所有行合并为一个长字符串，并全部转换为小写
    full_text = " ".join(text_lines).lower()
    
    # 需要清理的标点符号
    punctuations = ",.;:'\"?!"
    for p in punctuations:
        full_text = full_text.replace(p, " ")
    
    # 按照空格切分成单词列表
    words = full_text.split()

    #  3. 统计词频与停用词过滤 
    word_counts = {}
    for word in words:
        # 如果不是停用词，则加入字典进行统计
        if word not in stop_words:
            word_counts[word] = word_counts.get(word, 0) + 1

    #  4. 排序与输出
    # 将字典转为元组列表: [('word', count), ...]
    items = list(word_counts.items())
    
    # 排序规则：先按词频降序(-x[1])，词频相同时按单词字母升序(x[0])
    items.sort(key=lambda x: (-x[1], x[0]))

    # 获取 Top K (如果总词数不足 K，切片不会报错，会返回全部)
    top_k_items = items[:k]

    # 计算输出统计信息
    valid_word_total = sum(word_counts.values())
    unique_word_total = len(word_counts)

    # 打印结果
    print("\n" + "="*20 + " 输出结果 " + "="*20)
    print(f"有效单词总数：{valid_word_total}")
    print(f"不重复单词数量：{unique_word_total}")
    print(f"Top {k}：")
    for word, count in top_k_items:
        print(f"{word} : {count}")

# 执行主函数
if __name__ == "__main__":
    process_text()
```

# 为什么在编程时，选择合适的数据结构很重要？
1. **可读性**：结构选对了，代码更直观。  
   比如这段代码里：
   - 停用词用 `set`，表达“判断在不在集合里”；
   - 词频用 `dict`，表达“单词 → 次数”；
   - 排序结果用 `list` 存 `(word, count)`，方便排序和取 Top K。

2. **查找方式**：不同结构适合不同操作。  
   - `set` 适合频繁判断“是否存在”，所以过滤停用词快；
   - `dict` 适合按键更新计数，所以统计词频自然；
   - `list` 适合最终排序和切片。

3. **数据规模**：小数据看不出差别，大数据会放大问题。  
   几百行文本时用列表硬找停用词也可能很快；但停用词几千个、文本几十万行时，不合适的数据结构会明显变慢。

4. **运行效率**：选对结构能避免不必要的线性扫描和重复操作。  
   这段代码整体大致是“预处理 + 统计 + 排序”的开销；`set` 和 `dict` 让过滤和统计更高效，最后只排序一次。

合适的数据结构让代码更清楚、操作更匹配、规模上来后更稳，也更容易写对和维护

```python
# 1.3 文件与数据处理——活动数据清洗与 JSON 报告
import csv
import json
from pathlib import Path
from collections import Counter

def process_activity_data(input_file, output_file):
    # 确保输出目录存在（Step 5）
    output_path = Path(output_file)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    valid_records = []
    invalid_records = [] # 用于附加思考：记录无效行及原因

    try:
        with open(input_file, mode='r', encoding='utf-8') as f:
            reader = csv.DictReader(f)
            
            # 检查表头是否完整（Step 5）
            expected_fields = {'name', 'category', 'interest_score'}
            if not expected_fields.issubset(reader.fieldnames or []):
                raise ValueError(f"CSV表头缺失，必须包含: {expected_fields}")

            for row_num, row in enumerate(reader, start=2): # 从第2行开始（跳过表头）
                name = row.get('name', '').strip()
                category = row.get('category', '').strip()
                score_str = row.get('interest_score', '').strip()

                # 记录无效原因
                invalid_reason = []

                # 1. 校验 name
                if not name:
                    invalid_reason.append("name为空")
                
                # 2. 校验 category
                if not category:
                    invalid_reason.append("category为空")
                
                # 3. 校验 interest_score (分值)
                score = None
                if not score_str:
                    invalid_reason.append("interest_score为空")
                else:
                    try:
                        score = float(score_str)
                        if not (0 <= score <= 100):
                            invalid_reason.append("interest_score不在0-100范围内")
                    except ValueError:
                        invalid_reason.append("interest_score无法转换为数字")

                # 判断记录是否有效
                if len(invalid_reason) == 0:
                    valid_records.append({
                        'name': name,
                        'category': category,
                        'interest_score': score
                    })
                else:
                    invalid_records.append({
                        'row_num': row_num,
                        'reason': "，".join(invalid_reason),
                        'data': row
                    })

    except FileNotFoundError:
        print(f"错误：找不到输入文件 {input_file}")
        return
    except UnicodeDecodeError:
        print(f"错误：文件编码不正确，请确保文件是UTF-8编码")
        return
    except csv.Error as e:
        print(f"错误：CSV文件解析失败 - {e}")
        return
    except ValueError as e:
        print(f"数据错误: {e}")
        return

    # Step 3: 统计有效数据
    valid_count = len(valid_records)
    invalid_count = len(invalid_records)
    
    # 处理没有任何有效记录的情况 (Step 5)
    if valid_count == 0:
        report_data = {
            "valid_count": 0,
            "invalid_count": invalid_count,
            "average_interest_score": None,
            "top_activity": {
                "score": None,
                "activities": []
            },
            "category_counts": {}
        }
    else:
        # 计算平均分 (保留两位小数)
        total_score = sum(r['interest_score'] for r in valid_records)
        average_score = round(total_score / valid_count, 2)

        # 计算最高分及并列活动
        max_score = max(r['interest_score'] for r in valid_records)
        top_activities = [r['name'] for r in valid_records if r['interest_score'] == max_score]
        
        # 统计每个 category 的有效活动数量 (字典序排序)
        category_counts = dict(Counter(r['category'] for r in valid_records))
        sorted_category_counts = dict(sorted(category_counts.items()))

        report_data = {
            "valid_count": valid_count,
            "invalid_count": invalid_count,
            "average_interest_score": average_score,
            "top_activity": {
                "score": round(max_score, 2),
                "activities": top_activities
            },
            "category_counts": sorted_category_counts
        }

    # Step 4: 输出 JSON 报告
    with open(output_path, mode='w', encoding='utf-8') as f:
        json.dump(report_data, f, ensure_ascii=False, indent=4)

    # 按照题目要求的格式输出到控制台
    print(f"有效记录数：{valid_count}")
    print(f"无效记录数：{invalid_count}")
    
    if valid_count > 0:
        print(f"平均兴趣分：{average_score:.2f}")
        print(f"最高兴趣分：{max_score:.2f}")
        # 使用顿号连接并列的活动名称
        print(f"最高分活动：{'、'.join(top_activities)}")
        print("分类统计：")
        for category, count in sorted_category_counts.items():
            print(f"{category} {count}")
            

# 运行函数
process_activity_data('data/activities.csv', 'output/activity_report.json')
```

# “附加思考”问题 (数据处理策略)
首尾空格：使用 .strip() 方法去除。原因：空格会导致“ technology”和“technology”被识别为两个不同的类别，影响统计准确性

大小写不一致：在读取时，统一使用 .lower() 转换为小写（或统一大写）。原因：同理，Art 和 art 应该被视为同一个类别

重复记录：可以使用集合（Set）或 pandas 的 drop_duplicates() 去重。原因：重复记录会虚增活动数量，拉偏平均分

字段或表头缺失：在代码开始时检查 reader.fieldnames，如果缺失直接抛出异常并提示用户修复文件。原因：缺少必需列会导致后续逻辑全部崩溃，提前拦截更安全

单位混乱：例如分数出现 "85分" 或者 "0.85"（百分制与小数制混用）。需要写规则进行清洗（如去除中文字符，乘以100等）。原因：不统一单位无法进行数学计算

编码问题：使用 encoding='utf-8' 打开文件。如果遇到报错，捕获 UnicodeDecodeError 并提示用户。原因：中文环境极易出现 GBK 和 UTF-8 的冲突

数值不符合实际：如分数是 -10 或 9999。代码中通过 0 <= score <= 100 进行过滤并标记为无效。原因：异常值（Outliers）会严重破坏平均分和最高分的统计结果
